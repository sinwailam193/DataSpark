### 📌 Pre-requisites
- Create your own Storage Account + Container
- Register the Container as an External Location (+ Create Storage Credential if required)
- Create a catalog called **test_catalog_1**

### 📌 Writing to an External Table

```python
df = spark.read.table("samples.accuweather.historical_hourly_metric")
```

```python
path = "your path here"

df.write.option("path", path).saveAsTable("test_catalog_1.default.historical_hourly_metric_ext")
```

### 📌 Dropping an External Table

```python
spark.sql("DROP TABLE test_catalog_1.default.historical_hourly_metric_ext")
```

### 📌 Registering an External Table backed by existing data

```python
spark.catalog.createTable(
    tableName = "test_catalog_1.default.historical_hourly_metric_ext",
    path = path,
    source = "delta"
)
```

### 📌 External tables can use the DELTA, CSV, JSON, AVRO, PARQUET, ORC and TEXT file formats

```python
csv_path = "your path here"
df.write.option("path", csv_path).format("csv").saveAsTable("test_catalog_1.default.historical_hourly_metric_ext_csv")
```
