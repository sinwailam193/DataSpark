### 📌 Creating a Python UDF

```python
# Creating a Python UDF
from pyspark.sql.types import IntegerType

@udf(IntegerType())
def get_length(x):
    return len(x)
```

```python
# Calling the UDF
df = spark.read.table("population_metrics.default.countries_consolidated")

df.withColumn("country_length", get_length(df.country)).display()
```
