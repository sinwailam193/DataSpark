### 📌 How to set Spark Configurations

You can set your configurations in:
- Notebooks
- Clusters (Clusters -> Advanced -> Spark Configs)
- Jobs (to be discussed later)

```python
# Notebook configs
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", 104857600)
spark.conf.set("spark.databricks.io.cache.enabled", "false")
```

```python
# Get all configs as a list of key-value pairs
spark.conf.getAll
```
